```python
# 1. 环境初始化与库导入
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

# 设置图表风格
plt.rcParams['figure.figsize'] = (15, 10)
plt.rcParams['figure.dpi'] = 150  # 降低DPI，避免部分环境内存问题
plt.rcParams['font.sans-serif'] = ['Arial Unicode MS', 'SimHei']
plt.rcParams['axes.unicode_minus'] = False

# 2. 加载数据
df = pd.read_csv("/Users/xiahanfei/Desktop/MGS3001/MGS3001_Assignment3/data/FINAL_ANALYSIS_DATA.csv")

print("=== 数据基本信息 ===")
print(f"数据规模：{df.shape[0]} 行 × {df.shape[1]} 列")

# 3. 选择核心变量（从你的列表里挑6个有代表性的）
core_vars = [
    "Audio_AIGC_Rate", 
    "Lyrics_AIGC_Rate", 
    "Total_AI_Permeability", 
    "comment_count", 
    "Novelty", 
    "Usefulness"
]

# 过滤出存在的变量（防止拼写错误）
core_vars = [var for var in core_vars if var in df.columns]
n_vars = len(core_vars)

# 4. 创建子图
rows = 2 if n_vars >= 4 else 1
cols = 3 if n_vars >= 3 else n_vars
fig, axes = plt.subplots(rows, cols, squeeze=False)
axes = axes.flatten()

# 5. 绘制直方图
for idx, var in enumerate(core_vars):
    ax = axes[idx]
    data = df[var].dropna()
    
    ax.hist(data, bins=20, alpha=0.7, edgecolor='black', color='#1f77b4')
    ax.set_title(f'Distribution of {var}\n(Valid Values: {len(data)})', fontsize=10, fontweight='bold')
    ax.set_xlabel(var, fontsize=8)
    ax.set_ylabel('Frequency', fontsize=8)
    ax.grid(True, alpha=0.3, axis='y')
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

# 隐藏多余子图
for idx in range(n_vars, rows*cols):
    axes[idx].axis('off')

# 调整布局并显示图表
plt.tight_layout(pad=3.0)
plt.show()  # 只显示，不保存，避开所有路径/权限问题

print("\n✅ 图表已在上方显示，无需保存文件！")
```

```python
# 【终极版 · 一步到位】
# 直接读取你的文件 → 自动生成阶梯热力图 → 无对角线
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# 1. 直接读取你的数据（用你给的路径）
df = pd.read_csv('/Users/xiahanfei/Desktop/MGS3001/AI_music/FINAL_ANALYSIS_DATA.csv')

# 2. 选择你要的变量（和你截图完全一致）
heatmap_vars = [
    "Audio_AIGC_Rate", 
    "Lyrics_AIGC_Rate", 
    "Total_AI_Permeability", 
    "comment_count", 
    "Novelty", 
    "Usefulness"
]

# 3. 计算相关系数
corr_matrix = df[heatmap_vars].corr()

# --------------------------
# 核心：去掉对角线 + 阶梯形
# --------------------------
mask = np.triu(np.ones_like(corr_matrix, dtype=bool), k=0)

# 4. 画图（和你风格完全一样）
plt.figure(figsize=(10, 8))
sns.heatmap(
    corr_matrix,
    mask=mask,        # 关键：隐藏对角线与上三角
    annot=True,
    fmt='.2f',
    cmap='RdBu_r',
    center=0,
    vmin=-1,
    vmax=1,
    linewidths=0.5,
    linecolor='white',
    square=True,
    cbar_kws={'shrink': 0.8}
)

plt.title('Heat figure', fontsize=14, fontweight='bold')
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 读取你的文件
df = pd.read_csv('/Users/xiahanfei/Desktop/MGS3001/AI_music/filtered_data.csv')

# 要画的三个变量
vars_to_plot = ["Novelty", "Usefulness", "Total_AI_Permeability"]

for var in vars_to_plot:
    fig, ax = plt.subplots(figsize=(8, 4))

    # 绘制直方图 + KDE曲线
    sns.histplot(df[var], bins=15, kde=True, color="#3B82F6",
                 edgecolor="white", alpha=0.3, ax=ax)

    # 画均值线
    mean_val = df[var].mean()
    ax.axvline(mean_val, color="#DC2626", linestyle="--", linewidth=1.5,
               label=f"Mean = {mean_val:.2f}")

    # 设置标题和标签
    ax.set_title(f"Distribution of {var}",
                 fontsize=13, fontweight="bold")
    ax.set_xlabel(var)
    ax.set_ylabel("Frequency")
    ax.legend()
    sns.despine()
    plt.tight_layout()
    plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# 二次多项式拟合（检验U型/倒U型）
plt.figure(figsize=(12,5))

# 1. AI渗透率 ~ Novelty 非线性
plt.subplot(1,2,1)
sns.regplot(x="Total_AI_Permeability", y="Novelty", 
            data=df, order=2,  # order=2 就是二次曲线，检验U型
            color="#3B82F6", line_kws={"color":"red"})
plt.title("AI Permeability vs Novelty (Nonlinear Curve)")

# 2. AI渗透率 ~ Usefulness 非线性
plt.subplot(1,2,2)
sns.regplot(x="Total_AI_Permeability", y="Usefulness", 
            data=df, order=2,
            color="#3B82F6", line_kws={"color":"red"})
plt.title("AI Permeability vs Usefulness (Nonlinear Curve)")

plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

# 1次 线性
sns.regplot(x="Total_AI_Permeability", y="Novelty",
            data=df, order=1, scatter_kws={"alpha":0.5},
            label="Linear (1st order)", color="#3B82F6")

# 2次 U型/倒U
sns.regplot(x="Total_AI_Permeability", y="Novelty",
            data=df, order=2, scatter_kws={"alpha":0.0},
            label="Quadratic (2nd order)", color="#DC2626")

# 3次 门槛/波浪/分段曲线
sns.regplot(x="Total_AI_Permeability", y="Novelty",
            data=df, order=3, scatter_kws={"alpha":0.0},
            label="Cubic (3rd order)", color="#10B981")

plt.legend()
plt.title("AI Permeability vs Novelty \nLinear / Quadratic / Cubic Fit", fontsize=13)
plt.xlabel("Total AI Permeability")
plt.ylabel("Novelty")
plt.show()
```
