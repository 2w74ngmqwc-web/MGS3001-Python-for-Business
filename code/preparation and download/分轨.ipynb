```python
# -------------------------- 第一步：先装所有依赖 --------------------------
import os
import sys
print("📦 正在安装依赖...")
os.system('pip3 install -q pandas numpy transformers torch librosa')

# -------------------------- 第二步：导入工具，加载检测模型 --------------------------
import numpy as np
import pandas as pd
from transformers import pipeline, AutoModelForSequenceClassification, AutoTokenizer

print("🤖 正在加载检测模型...")
# 1. 歌词AIGC检测模型（公开的，没问题）
text_tokenizer = AutoTokenizer.from_pretrained("Hello-SimpleAI/chatgpt-detector-roberta")
text_model = AutoModelForSequenceClassification.from_pretrained("Hello-SimpleAI/chatgpt-detector-roberta")
text_detector = pipeline("text-classification", model=text_model, tokenizer=text_tokenizer)

# 2. 音频AIGC检测模型！换成了这个完全公开的！
# 专门检测AI生成的音频/音乐，所有人都能下载，不需要权限！
audio_detector = pipeline("audio-classification", model="MattyB95/AST-AMVD-SAD-v1")

# -------------------------- 第三步：路径配置（已经根据你的截图改好了） --------------------------
BASE_DIR = '/Users/xiahanfei/Desktop/MGS3001/AI_music/ Control group/ai_music_downloads_original'
AUDIO_DIR = os.path.join(BASE_DIR, 'audio')
LYRIC_DIR = os.path.join(BASE_DIR, 'lyrics')
SEPARATED_DIR = '/Users/xiahanfei/Desktop/MGS3001/AI music/experimental group/AIGC rate/separated/mdx_extra_q/'
CSV_PATH = os.path.join(BASE_DIR, 'novelty_usefulness_results.csv')

# 我们研究的权重配置
WEIGHTS = {
    'idea': 0.15,
    'lyric': 0.15,
    'melody': 0.20,
    'harmony': 0.15,
    'arrangement': 0.15,
    'vocal': 0.10,
    'mix': 0.10
}

# -------------------------- 第四步：自动处理每一首歌 --------------------------
def process_single_song(song_name):
    print(f"\n🎵 正在计算 {song_name} 的AI渗透率...")
    
    # 1. 读取你已经拆分好的音轨
    vocal_path = os.path.join(SEPARATED_DIR, song_name, 'vocals.wav')
    accomp_path = os.path.join(SEPARATED_DIR, song_name, 'other.wav')
    lyric_path = os.path.join(LYRIC_DIR, f'{song_name}.txt')
    
    # 检查文件是否存在
    if not os.path.exists(vocal_path):
        print(f"  ⚠️  找不到人声音轨: {vocal_path}")
        return None
    if not os.path.exists(accomp_path):
        print(f"  ⚠️  找不到伴奏音轨: {accomp_path}")
        return None
    if not os.path.exists(lyric_path):
        print(f"  ⚠️  找不到歌词文件: {lyric_path}")
        return None
    
    # 2. 分环节检测
    contributions = {}
    
    # 检测歌词是不是AI写的
    print("  📝 检测歌词...")
    with open(lyric_path, 'r', encoding='utf-8') as f:
        lyric = f.read()
    if len(lyric) > 0:
        res = text_detector(lyric[:512])[0]
        contributions['lyric'] = 1.0 if res['label'] == 'chatgpt' else 0.0
    else:
        contributions['lyric'] = 0.0
    
    # 检测人声是不是AI生成的
    print("  🎤 检测人声...")
    vocal_res = audio_detector(vocal_path)
    # 这个模型的label："AI Generated Audio" 就是AI生成的
    contributions['vocal'] = 1.0 if 'AI Generated' in vocal_res[0]['label'] else 0.0
    
    # 检测伴奏/编曲是不是AI生成的
    print("  🎹 检测伴奏/编曲...")
    accomp_res = audio_detector(accomp_path)
    contributions['arrangement'] = 1.0 if 'AI Generated' in accomp_res[0]['label'] else 0.0
    
    # 其他环节：全AI歌的这些环节都是AI做的
    contributions['melody'] = contributions['arrangement']
    contributions['harmony'] = contributions['arrangement']
    contributions['mix'] = contributions['arrangement']
    contributions['idea'] = 1.0
    
    # 3. 加权计算最终的AI渗透率
    permeability = sum(
        contributions[dim] * weight 
        for dim, weight in WEIGHTS.items()
    )
    print(f"  ✅ {song_name} 的AI渗透率: {permeability:.2f}")
    return permeability

# -------------------------- 第五步：批量处理，更新CSV --------------------------
def main():
    # 读取所有歌曲
    song_names = [os.path.splitext(f)[0] for f in os.listdir(AUDIO_DIR) if f.endswith(('.mp3', '.wav'))]
    print(f"🔍 共找到 {len(song_names)} 首歌待处理")
    
    # 读取原来的CSV
    if os.path.exists(CSV_PATH):
        df = pd.read_csv(CSV_PATH)
    else:
        df = pd.DataFrame(columns=['song_name'])
    
    # 逐个处理
    for song_name in song_names:
        # 如果已经处理过了就跳过
        if song_name in df['song_name'].values and 'ai_permeability' in df.columns and not pd.isna(df[df['song_name']==song_name]['ai_permeability'].values[0]):
            print(f"⏭️  {song_name} 已经处理过了，跳过")
            continue
        
        # 计算渗透率
        p = process_single_song(song_name)
        if p is None:
            continue
        
        # 更新到df里
        if song_name in df['song_name'].values:
            df.loc[df['song_name']==song_name, 'ai_permeability'] = p
        else:
            new_row = pd.DataFrame({'song_name': [song_name], 'ai_permeability': [p]})
            df = pd.concat([df, new_row], ignore_index=True)
    
    # 保存更新后的CSV
    df.to_csv(CSV_PATH, index=False)
    print(f"\n🎉 全部处理完成！")
    print(f"💾 结果已经更新到你的CSV文件: {CSV_PATH}")
    print(f"📊 现在你可以运行之前的合并数据、画图的代码，就能得到完整的倒U型曲线了！")

if __name__ == "__main__":
    main()
```
